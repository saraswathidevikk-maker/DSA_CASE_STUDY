```python
import numpy as np
import pandas as pd
import json
```

```python
# Task 1 - Read the Population dataset from the CSV file.
df1 = pd.read_csv("/content/population.csv")
print("Shape of dataset:")
print(df1.shape)
print("Data types:")
print(df1.dtypes)
df1.head()
```

```python
# Task 2 - Read the GDP dataset from the Excel file.
df2 = pd.read_excel("/content/gdp.xlsx")
print("Data types:")
print(df2.dtypes)
df2.head()
```

```python
# Task 3 - Read the Internet Users dataset from the JSON file.
with open("/content/internet_users.json", "r") as file:
    data = json.load(file)
df3 = pd.DataFrame(data)
df3.info()
df3.head()
```

```python
# Task 4 - Read the Literacy Rate dataset from the XML file.
df4 = pd.read_xml("/content/literacy_rate.xml")
df4.head()
```

```python
# Part B - Data Understanding - Task 5 - Explore each dataset.
#population.csv

print("Number of rows:", df1.shape[0])
print("Number of columns:", df1.shape[1])

print("Column names:")
print(df1.columns)

print("\nData types:")
print(df1.dtypes)

print("\nMissing values:")
print(df1.isnull().sum())

print("\nDuplicate records:")
print(df1.duplicated().sum())
```

```python
#gdp.xlsx

print("Number of rows:", df2.shape[0])
print("Number of columns:", df2.shape[1])

print("Column names:")
print(df2.columns)

print("\nData types:")
print(df2.dtypes)

print("\nMissing values:")
print(df2.isnull().sum())

print("\nDuplicate records:")
print(df2.duplicated().sum())
```

```python
#internet_users.json

print("Number of rows:", df3.shape[0])
print("Number of columns:", df3.shape[1])

print("Column names:")
print(df3.columns)

print("\nData types:")
print(df3.dtypes)

print("\nMissing values:")
print(df3.isnull().sum())

print("\nDuplicate records:")
print(df3.duplicated().sum())
```

```python
#literacy_rate.xml

print("Number of rows:", df4.shape[0])
print("Number of columns:", df4.shape[1])

print("Column names:")
print(df4.columns)

print("\nData types:")
print(df4.dtypes)

print("\nMissing values:")
print(df4.isnull().sum())

print("\nDuplicate records:")
print(df4.duplicated().sum())
```

```python
# Part C – Data Transformation - Task 6 Rename columns wherever necessary so that all datasets have a common column name for merging.
df1.rename(columns={'Country Name': 'country'}, inplace=True)

df2.rename(columns={
    'Country': 'country',
    'GDP': 'gdp'
}, inplace=True)

df3.rename(columns={'Country Name': 'country'}, inplace=True)

df4.rename(columns={
    'name': 'country',
    'Literacy Rate': 'literacy_rate'
}, inplace=True)
```

```python
# Task 7 - Convert the required columns into appropriate numeric data types.
df2.info()

df2['gdp'] = pd.to_numeric(df2['gdp'], errors='coerce')

print(df2.dtypes)
```

```python
df4.info()
df4['literacy_rate'] = pd.to_numeric(
    df4['literacy_rate'], errors='coerce')
```

```python
df3.info()
df3['internet_users'] = pd.to_numeric(
    df3['internet_users'], errors='coerce')
```

```python
df1.info()
df1['population'] = pd.to_numeric(
    df1['population'], errors='coerce')
```

```python
# Task 8 - Merge all four datasets into one DataFrame.
finaldf = pd.merge(df1, df2, on='country', how='outer')
finaldf = pd.merge(finaldf, df3, on='country', how='outer')
finaldf = pd.merge(finaldf, df4, on='country', how='outer')
finaldf.head()
```

```python
# Task 9 - Create the following calculated columns: 1. Internet Penetration (%) 2. GDP Per Capita.
finaldf['internet_penetration'] = ( finaldf['internet_users'] / finaldf['population']) * 100
finaldf['internet_penetration'] = finaldf[ 'internet_penetration'].round(2)
finaldf.head()
```

```python
# Part D – Load Data - Task 10 Create a SQLite database. Store the final merged dataset into a table named ‘country_statistics’.
import sqlite3
conn = sqlite3.connect('/content/country.db')
finaldf.to_sql(
    'country_statistics',
    conn,
    if_exists='replace',
    index=False
    )
print("Data successfully stored in database")
```

```python
# Task 11 - Verify that the data has been loaded successfully.
query = "SELECT * FROM country_statistics LIMIT 5"
result = pd.read_sql(query, conn)
result
```

```python
# Part E – Data Analysis Using Pandas.
# List the Top 5 countries by GDP.
finaldf.nlargest(5, 'gdp')[['country', 'gdp']]
```

```python
# List the Top 5 countries by population.
finaldf.nlargest(5, 'population')[['country', 'population']]
```

```python
# Calculate the average literacy rate
finaldf['literacy_rate'].mean()
```

```python
# List all countries having a literacy rate greater than 90%.
finaldf[finaldf['literacy_rate'] > 90][['country', 'literacy_rate']]
```

```python
# List all countries having internet penetration greater than 70%.
finaldf[finaldf['internet_users'] > 70][['country', 'internet_users']]
```

```python
# List the Top 5 countries by GDP per capita.
finaldf['gdp_per_capita'] = finaldf['gdp'] / finaldf['population']
finaldf.nlargest(5, 'gdp_per_capita')[['country', 'gdp_per_capita']]
```

```python
# Calculate the total population represented in the dataset.
finaldf['population'].sum()
```

```python
# Calculate the average internet penetration across all countries.
finaldf['internet_users'].mean()
```

```python
# Identify the country having the highest number of internet users.
finaldf.nlargest(1, 'internet_users')[['country', 'internet_users']]
```

```python
finaldf['number_of_internet_users'] = (
    finaldf['population'] * finaldf['internet_users'] / 100
)

finaldf.nlargest(1, 'number_of_internet_users')[
    ['country', 'number_of_internet_users']
]
```

```python
# Find the correlation between Literacy Rate and Internet Penetration.
finaldf['literacy_rate'].corr(finaldf['internet_users'])
```

```python
# Part F – Data Analysis Using SQL.
# Display the Top 5 countries by GDP
query = "SELECT country, gdp FROM country_statistics ORDER BY gdp DESC LIMIT 5"
result = pd.read_sql(query, conn)
result
```

```python
# Find the average literacy rate.
query = "SELECT AVG(literacy_rate) FROM country_statistics"
result = pd.read_sql(query, conn)
result
```

```python
# Display all countries having literacy rate above 90%.
query = "SELECT country FROM country_statistics WHERE literacy_rate > 90"
result = pd.read_sql(query, conn)
result
```

```python
# Find the Top 5 countries with the highest internet penetration.
query = """
SELECT country, internet_penetration FROM country_statistics
ORDER BY internet_penetration DESC
LIMIT 5
"""
pd.read_sql(query, conn)
```

```python
# Calculate the total population.
query = """
SELECT SUM(population) AS total_population
FROM country_statistics
"""
pd.read_sql(query, conn)
```

```python
# Find the country with the highest GDP per capita.
query = """
SELECT country,
       gdp * 1.0 / population AS gdp_per_capita
FROM country_statistics
WHERE population > 0
ORDER BY gdp_per_capita DESC
LIMIT 1
"""
pd.read_sql(query, conn)
```

```python
#  Export the final merged dataset as a CSV file.
finaldf.to_csv('finaldf.csv', index=False)
print("CSV file exported successfully")
```

```python
# Sort the dataset in descending order of GDP per Capita and display the Top 10 countries.
query = """
SELECT
    country,
    gdp,
    population,
    gdp * 1.0 / population AS gdp_per_capita
FROM country_statistics
WHERE population > 0
ORDER BY gdp_per_capita DESC
LIMIT 10
"""
pd.read_sql(query, conn)
```
