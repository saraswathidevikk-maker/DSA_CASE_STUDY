# QuickCart – SQL Case Study

## 1. Connect to the SQLite database
Keep `QuickCartDB.db` in the same folder as this notebook.

```python
import sqlite3
import pandas as pd

db_path = "QuickCart_DB.db"
conn = sqlite3.connect(db_path)
conn.execute("PRAGMA foreign_keys = ON;")
print("Connected to QuickCart!")
```

## 2. Explore the database
List the available tables.

```python
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';", conn
)
display(tables)
```

```python
for table in ["Customers", "Products", "Orders", "Order_Items", "Payments"]:
    print("\\n---", table, "---")
    display(pd.read_sql_query(f"PRAGMA table_info({table});", conn))
```

```python
for table in ["Customers", "Products", "Orders", "Order_Items", "Payments"]:
    print("\\n---", table, "---")
    display(pd.read_sql_query(f"SELECT * FROM {table} LIMIT 5;", conn))
```

Mission 2 – Understand sales performance

```python
# Find the highest-priced product.
query1 = """
SELECT product_name, category, unit_price FROM Products
WHERE unit_price = (SELECT MAX(unit_price) FROM Products);
"""
pd.read_sql_query(query1, conn)
```

```python
# Find the lowest-priced product.
query2 = """
SELECT product_name, category, unit_price FROM Products
WHERE unit_price = (SELECT MIN(unit_price) FROM Products);
"""

pd.read_sql_query(query2, conn)
```

```python
# Calculate the total quantity of products sold.
query3 = """
SELECT SUM(quantity) AS Total_Quantity_Sold FROM Order_Items;
"""
pd.read_sql_query(query3, conn)
```

```python
# Calculate the total sales value.
query4 = """
SELECT SUM(quantity * selling_price) AS Total_Sales_Value
FROM Order_Items;
"""
pd.read_sql_query(query4, conn)
```

```python
# Calculate the average selling price of products sold
query5 = """
SELECT AVG(selling_price) AS Average_Selling_Price
FROM Order_Items;
"""

pd.read_sql_query(query5, conn)
```

```python
# Find the number of orders for each order status.
query6 = """
SELECT order_status, COUNT(order_id) AS Total_Orders
FROM Orders
GROUP BY order_status;
"""
pd.read_sql_query(query6, conn)
```

**Mission 3 – Customer behaviour analysis**


```python
# Display each order along with the customer name.
query1 = """ SELECT o.order_id, c.customer_name, o.order_date, o.order_status FROM Orders o INNER JOIN Customers c ON o.customer_id = c.customer_id;"""
pd.read_sql_query(query1, conn)
```

```python
# Find the number of orders placed by each customer.
query2 = """ SELECT c.customer_name, COUNT(o.order_id) AS total_orders FROM Customers c LEFT JOIN Orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_name;"""
pd.read_sql_query(query2, conn)
```

```python
# Calculate the total amount spent by each customer.
query3 = """ SELECT c.customer_name, SUM(oi.quantity * oi.selling_price) AS total_spent FROM Customers c JOIN Orders o ON c.customer_id = o.customer_id JOIN Order_Items oi ON o.order_id = oi.order_id GROUP BY c.customer_name
ORDER BY total_spent DESC;"""
pd.read_sql_query(query3, conn)
```

```python
# Calculate the total amount spent by each customer.
query3 = """ SELECT c.customer_name, SUM(oi.quantity * oi.selling_price) AS total_spent FROM Customers c JOIN Orders o ON c.customer_id = o.customer_id JOIN Order_Items oi ON o.order_id = oi.order_id GROUP BY c.customer_name
ORDER BY total_spent DESC;"""
pd.read_sql_query(query3, conn)
```

```python
# Identify the customer who has spent the most.
query4 = """ SELECT c.customer_name, SUM(oi.quantity * oi.selling_price) AS total_spent FROM Customers c JOIN Orders o ON c.customer_id = o.customer_id JOIN Order_Items oi ON o.order_id = oi.order_id GROUP BY c.customer_id, c.customer_name
ORDER BY total_spent DESC
LIMIT 1;
"""
pd.read_sql_query(query4, conn)
```

```python
# Find customers who have never placed an order.
query5 = """ SELECT c.customer_name FROM Customers c LEFT JOIN Orders o ON c.customer_id = o.customer_id WHERE o.order_id IS NULL;"""
pd.read_sql_query(query5, conn)
```

```python
# Find customers who have placed more than two orders.
query6 = """ SELECT c.customer_name, COUNT(o.order_id) AS total_orders FROM Customers c JOIN Orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
HAVING COUNT(o.order_id) > 2;
"""
pd.read_sql_query(query6, conn)
```

```python
# Find the average order value.
query7 = """ SELECT AVG(oi.quantity * oi.selling_price) AS average_order_value FROM Order_Items oi;"""
pd.read_sql_query(query7, conn)
```

```python
# Identify the city with the highest number of customers.
query8 = """ SELECT city, COUNT(*) AS total_customers FROM Customers GROUP BY city
ORDER BY total_customers DESC
LIMIT 1;
"""
pd.read_sql_query(query8, conn)
```

```python
# Find customers who registered before a specified date.
query9 = """ SELECT * FROM Customers
WHERE registration_date < '2025-01-01';
"""
pd.read_sql_query(query9, conn)
```

```python
# Identify repeat customers.
query10 = """ SELECT c.customer_name, COUNT(o.order_id) AS total_orders FROM Customers c JOIN Orders o
ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
HAVING COUNT(o.order_id) > 1;
"""
pd.read_sql_query(query10, conn)
```

**Mission 4 – Payment and order investigation**

```python
#Display orders with their corresponding payment status.
query1 = """
SELECT orders.order_id, orders.order_date, orders.order_status,
payments.payment_method, payments.payment_status,payments.amount
FROM Orders
INNER JOIN Payments
ON orders.order_id = payments.order_id;
"""
pd.read_sql_query(query1, conn)
```

```python
#Find all unpaid orders
query2 = """
SELECT orders.order_id, orders.order_status, payments.payment_status, payments.amount
FROM Orders  INNER JOIN Payments  ON orders.order_id = payments.order_id
WHERE payments.payment_status = 'Unpaid';
"""

pd.read_sql_query(query2, conn)
```

```python
#Find all cancelled orders.
query3 = """ SELECT * FROM Orders WHERE order_status = 'Cancelled'; """

pd.read_sql_query(query3, conn)
```

```python
#Calculate the total amount received through each payment method.
query4 = """
SELECT payment_method, SUM(amount) AS Total_Amount_Received
FROM Payments WHERE payment_status = 'Paid'
GROUP BY payment_method;
"""

pd.read_sql_query(query4, conn)
```

```python
#Count successful and failed payments.
query5 = """
SELECT payment_status, COUNT(payment_id) AS Total_Payments
FROM Payments GROUP BY payment_status;
"""

pd.read_sql_query(query5, conn)
```

```python
#Find orders where the order status is Delivered but the payment status is not Paid.
query6 = """
SELECT orders.order_id, orders.order_status, payments.payment_status, payments.amount
FROM Orders  INNER JOIN Payments  ON orders.order_id = payments.order_id
WHERE orders.order_status = 'Delivered' AND payments.payment_status != 'Paid';
"""

pd.read_sql_query(query6, conn)
```

```python
#Identify the most commonly used payment method.
query7 = """
SELECT payment_method, COUNT(payment_id) AS Total_Payments
FROM Payments GROUP BY payment_method
ORDER BY Total_Payments DESC
LIMIT 1;
"""

pd.read_sql_query(query7, conn)
```

Mission 5 – Management challenge

```python
# Which product category should receive more attention from the marketing team?
query1 = """ SELECT p.category, SUM(oi.quantity * oi.selling_price) AS total_sales FROM Products p JOIN Order_Items oi
ON p.product_id = oi.product_id
GROUP BY p.category
ORDER BY total_sales ASC;
"""
pd.read_sql_query(query1, conn)
```

```python
# Which products need restocking?
query2 = """ SELECT product_id, product_name, stock_quantity FROM Products
WHERE stock_quantity < 10
ORDER BY stock_quantity ASC;
"""
pd.read_sql_query(query2, conn)
```

```python
# Which payment method is most popular?
query3 = """ SELECT payment_method, COUNT(*) AS total_payments FROM Payments
GROUP BY payment_method
ORDER BY total_payments DESC;
"""
pd.read_sql_query(query3, conn)
```

```python
# Find the second-highest-priced product.
query4 = """ SELECT product_name, unit_price FROM Products
ORDER BY unit_price DESC
LIMIT 1 OFFSET 1;
"""
pd.read_sql_query(query4, conn)
```

```python
# Find products that have never been ordered.
query5 = """ SELECT p.product_name FROM Products p LEFT JOIN Order_Items oi
ON p.product_id = oi.product_id
WHERE oi.order_id IS NULL;
"""
pd.read_sql_query(query5, conn)
```

```python
# Find orders containing more than one product line.
query6 = """ SELECT order_id, COUNT(*) AS total_product_lines
FROM Order_Items
GROUP BY order_id
HAVING COUNT(*) > 1;
"""
pd.read_sql_query(query6, conn)
```
